# ClimateGuard — Live Fire and Sea-Level APIs

```python
import os, requests, pandas as pd
from dotenv import load_dotenv
load_dotenv('../.env')
lat, lon = 19.08, 72.88
key=os.getenv('NASA_FIRMS_MAP_KEY')
if key:
    url=f'https://firms.modaps.eosdis.nasa.gov/api/area/csv/{key}/VIIRS_SNPP_NRT/{lon-3},{lat-3},{lon+3},{lat+3}/1'
    r=requests.get(url,timeout=30); r.raise_for_status()
    fires=pd.read_csv(pd.io.common.StringIO(r.text)); print(fires.head(), 'count=',len(fires))
else:
    print('Set NASA_FIRMS_MAP_KEY to query live fire detections.')
```

```python
key=os.getenv('STORMGLASS_API_KEY')
if key:
    now=pd.Timestamp.utcnow(); end=now+pd.Timedelta(hours=24)
    r=requests.get('https://api.stormglass.io/v2/weather/point',params={'lat':lat,'lng':lon,'params':'seaLevel,waveHeight,waterTemperature','start':now.isoformat(),'end':end.isoformat()},headers={'Authorization':key},timeout=20)
    r.raise_for_status(); print(r.json().get('hours',[])[:5])
else:
    print('Set STORMGLASS_API_KEY to query live sea-level/marine data.')
```
