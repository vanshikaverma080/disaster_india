# ClimateGuard — Live Hazard Feeds
This notebook uses live APIs only. No bundled data files are required.

Configure `NASA_FIRMS_MAP_KEY` and `STORMGLASS_API_KEY` in the Django `.env`. USGS earthquakes require no key.

```python
import os, requests
from dotenv import load_dotenv
load_dotenv('../.env')

lat, lon = 28.61, 77.21
usgs = requests.get('https://earthquake.usgs.gov/fdsnws/event/1/query', params={'format':'geojson','latitude':lat,'longitude':lon,'maxradiuskm':1200,'minmagnitude':2.5,'limit':20}, timeout=20).json()
print('Earthquakes:', len(usgs.get('features', [])))

firms_key=os.getenv('NASA_FIRMS_MAP_KEY')
print('NASA FIRMS configured:', bool(firms_key))
print('Stormglass configured:', bool(os.getenv('STORMGLASS_API_KEY')))
```

```python
# Quick inspect of recent earthquake events
[(f['properties'].get('mag'), f['properties'].get('place')) for f in usgs.get('features', [])[:10]]
```
