# ClimateGuard — Live Multi-Hazard Prediction
The Django API exposes live hazard snapshots. This notebook demonstrates how to consume them instead of local datasets.

```python
import requests, pandas as pd
base='http://127.0.0.1:8000/api'
district='Mumbai'
r=requests.get(f'{base}/hazards/{district}',timeout=30); r.raise_for_status(); x=r.json()
print({k:x.get(k) for k in ['district','flood','earthquake','fire','sealevel']})
```

```python
# Flood prediction using live weather
r=requests.post(f'{base}/predict/',json={'district':district,'use_live':True},timeout=30); r.raise_for_status(); print(r.json())
```
